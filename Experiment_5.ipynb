```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
from IPython.display import display

sns.set_theme(style="whitegrid", palette="viridis")
plt.rcParams["figure.figsize"] = (10, 5)
plt.rcParams["figure.dpi"] = 110

from google.colab import drive
drive.mount('/content/drive')

df = pd.read_csv("/content/drive/MyDrive/WI 7th Sem Practical Codes/Online Retail.csv", encoding="latin1")

print("DATASET OVERVIEW")
print("Original dataset shape:", df.shape)
display(df.head())

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"])
df["CustomerID"] = pd.to_numeric(df["CustomerID"], errors="coerce")
df = df.dropna(subset=["CustomerID", "Description"])
df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))].copy()
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
print("\nDATA AFTER CLEANING")
print("Cleaned dataset shape:", df.shape)
print("Unique customers:", df["CustomerID"].nunique())
print("Unique products:", df["Description"].nunique())
print("Total transactions:", df["InvoiceNo"].nunique())
print("Total revenue: £{:,.2f}".format(df["Revenue"].sum()))

top_products = df.groupby("Description")["Quantity"].sum().nlargest(10)
monthly_sales = df.set_index("InvoiceDate").resample("ME")["Revenue"].sum()
country_sales = df.groupby("Country")["Revenue"].sum().nlargest(10)

fig, axes = plt.subplots(2, 2, figsize=(16, 11))
sns.barplot(
    x=top_products.values, y=top_products.index,
    hue=top_products.index, palette="viridis",
    legend=False, ax=axes[0, 0])
axes[0, 0].set_title("Top 10 Best-Selling Products")
axes[0, 0].set_xlabel("Quantity Sold")
axes[0, 1].plot(monthly_sales.index, monthly_sales.values,
                marker="o", linewidth=2, color="#8e44ad")
axes[0, 1].fill_between(monthly_sales.index, monthly_sales.values,
                        alpha=0.15, color="#8e44ad")
axes[0, 1].set_title("Monthly Revenue Trend")
axes[0, 1].set_ylabel("Revenue (£)")
axes[0, 1].tick_params(axis="x", rotation=45)
sns.barplot(
    x=country_sales.values, y=country_sales.index,
    hue=country_sales.index, palette="magma",
    legend=False, ax=axes[1, 0])
axes[1, 0].set_title("Top 10 Countries by Revenue")
axes[1, 0].set_xlabel("Revenue (£)")
sns.heatmap(
    df[["Quantity", "UnitPrice", "Revenue"]].corr(),
    annot=True, cmap="coolwarm", fmt=".2f",
    linewidths=0.5, ax=axes[1, 1])
axes[1, 1].set_title("Feature Correlation Heatmap")
plt.suptitle("E-Commerce Data Analysis", fontsize=18, fontweight="bold")
plt.tight_layout()
plt.show()

print("\nASSOCIATION RULE MINING")

top_items = df["Description"].value_counts().head(30).index
basket_data = df[df["Description"].isin(top_items)]
transactions = basket_data.groupby("InvoiceNo")["Description"].apply(
    lambda x: list(set(x)))

encoder = TransactionEncoder()
encoded_data = encoder.fit(transactions).transform(transactions)
basket = pd.DataFrame(encoded_data, columns=encoder.columns_)

frequent_items = apriori(
    basket,
    min_support=0.01,
    use_colnames=True,
    max_len=3)
if not frequent_items.empty:
    rules = association_rules(
        frequent_items,
        metric="confidence",
        min_threshold=0.2)
    if not rules.empty:
        rules["Antecedents"] = rules["antecedents"].apply(
            lambda x: ", ".join(x))
        rules["Consequents"] = rules["consequents"].apply(
            lambda x: ", ".join(x))
        rules = rules.sort_values("lift", ascending=False)
        print("Frequent itemsets found:", len(frequent_items))
        print("Association rules found:", len(rules))
        display(
            rules[[
                "Antecedents", "Consequents",
                "support", "confidence", "lift"
            ]].head(10).round(3))

        top_rules = rules.head(10).copy()
        top_rules["Rule"] = (
            top_rules["Antecedents"] + " → " +
            top_rules["Consequents"])
        plt.figure(figsize=(11, 6))
        sns.barplot(
            data=top_rules, x="lift", y="Rule",
            hue="lift", palette="mako", legend=False)
        plt.title("Top Product Association Rules")
        plt.xlabel("Lift")
        plt.ylabel("Product Combinations")
        plt.tight_layout()
        plt.show()
    else:
        print("No association rules found. Try lowering the confidence threshold.")
else:
    print("No frequent itemsets found. Try lowering minimum support.")

print("\nCUSTOMER SEGMENTATION")
reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)
rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("Revenue", "sum"))
display(rfm.head())

rfm_scaled = np.log1p(rfm)
scaler = StandardScaler()
scaled_data = scaler.fit_transform(rfm_scaled)

kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)
rfm["Cluster"] = kmeans.fit_predict(scaled_data)

cluster_summary = rfm.groupby("Cluster").agg(
    Customers=("Recency", "count"),
    Avg_Recency=("Recency", "mean"),
    Avg_Frequency=("Frequency", "mean"),
    Avg_Monetary=("Monetary", "mean")).round(2)

print("\nCUSTOMER CLUSTER SUMMARY")
display(cluster_summary)

plt.figure(figsize=(10, 5))
sns.scatterplot(
    data=rfm, x="Frequency", y="Monetary",
    hue="Cluster", palette="viridis",
    s=65, alpha=0.7)
plt.title("Customer Segments")
plt.xlabel("Purchase Frequency")
plt.ylabel("Total Spending (£)")
plt.tight_layout()
plt.show()

print("\nBUSINESS INTELLIGENCE INSIGHTS")
print("1. Best-selling product:", top_products.index[0])
print("2. Highest revenue country:", country_sales.index[0])
print("3. Number of customer segments:", rfm["Cluster"].nunique())
best_cluster = cluster_summary["Avg_Monetary"].idxmax()
print("4. Highest spending customer cluster:", best_cluster)
```
