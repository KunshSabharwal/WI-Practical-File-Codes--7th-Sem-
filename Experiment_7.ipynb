```python
import pandas as pd
import matplotlib.pyplot as plt
from mlxtend.preprocessing import TransactionEncoder
from mlxtend.frequent_patterns import apriori, association_rules
from IPython.display import display
from google.colab import drive
drive.mount('/content/drive')
df = pd.read_csv("/content/drive/MyDrive/WI 7th Sem Practical Codes/Online Retail.csv", encoding="latin1")
print("Original Dataset Shape:", df.shape)

df["InvoiceDate"] = pd.to_datetime(df["InvoiceDate"], errors="coerce")
df = df.dropna(subset=["CustomerID", "Description", "InvoiceDate"])
df = df.drop_duplicates()
df = df[
    (df["Quantity"] > 0) &
    (df["UnitPrice"] > 0) &
    (~df["InvoiceNo"].astype(str).str.startswith("C"))].copy()
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
print("\nCleaned Dataset Shape:", df.shape)
print("Total Transactions:", df["InvoiceNo"].nunique())
print("Total Customers:", df["CustomerID"].nunique())
print("Total Revenue: £", round(df["Revenue"].sum(), 2))

top_products = df.groupby("Description")["Quantity"].sum().nlargest(10)
monthly_sales = df.set_index("InvoiceDate").resample("ME")["Revenue"].sum()

top_items = df["Description"].value_counts().head(30).index
transactions = df[df["Description"].isin(top_items)].groupby("InvoiceNo")["Description"].apply(
    lambda x: list(set(x)))
encoder = TransactionEncoder()
basket = pd.DataFrame(
    encoder.fit_transform(transactions),
    columns=encoder.columns_)
frequent_items = apriori(basket, min_support=0.01, use_colnames=True)
rules = association_rules(frequent_items, metric="confidence", min_threshold=0.2)
print("\nFrequent Itemsets Found:", len(frequent_items))
print("Association Rules Found:", len(rules))
if not rules.empty:
    rules["Antecedents"] = rules["antecedents"].apply(lambda x: ", ".join(sorted(x)))
    rules["Consequents"] = rules["consequents"].apply(lambda x: ", ".join(sorted(x)))
    rules_table = rules[
        ["Antecedents", "Consequents", "support", "confidence", "lift"]
    ].sort_values("lift", ascending=False).head(10)
    print("\nTop 10 Association Rules:")
    display(rules_table.round(3).reset_index(drop=True))
else:
    print("No association rules found. Try reducing the support threshold.")

reference_date = df["InvoiceDate"].max() + pd.Timedelta(days=1)
rfm = df.groupby("CustomerID").agg(
    Recency=("InvoiceDate", lambda x: (reference_date - x.max()).days),
    Frequency=("InvoiceNo", "nunique"),
    Monetary=("Revenue", "sum"))
print("\nRFM Analysis (First 5 Customers):")
display(rfm.head().round(2))
print("\nAverage RFM Values:")
display(rfm.mean().to_frame("Average").round(2))

plt.figure(figsize=(10, 5))
top_products.sort_values().plot(kind="barh")
plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Quantity Sold")
plt.ylabel("Product")
plt.tight_layout()
plt.show()
plt.figure(figsize=(10, 5))
monthly_sales.plot(marker="o")
plt.title("Monthly Revenue Trend")
plt.xlabel("Month")
plt.ylabel("Revenue (£)")
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```
