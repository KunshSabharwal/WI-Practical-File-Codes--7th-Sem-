```python
!pip -q install requests beautifulsoup4
import requests
from bs4 import BeautifulSoup
from collections import Counter
from urllib.parse import urljoin
url = "https://example.com"
headers = {"User-Agent": "Mozilla/5.0"}
response = requests.get(url, headers=headers, timeout=10)
if response.status_code == 200:
    soup = BeautifulSoup(response.text, "html.parser")
    print("=" * 55)
    print("        WEB ANALYTICS RESULT")
    print("=" * 55)
    title = soup.title.get_text(strip=True) if soup.title else "Not Available"
    print("\nPage Title:", title)
    headings = soup.find_all(["h1", "h2", "h3"])
    paragraphs = soup.find_all("p")
    links = soup.find_all("a", href=True)
    print("\nWEBPAGE CONTENT STATISTICS")
    print("-" * 40)
    print("Number of Headings   :", len(headings))
    print("Number of Paragraphs :", len(paragraphs))
    print("Number of Links      :", len(links))
    print("\nHEADINGS / POPULAR CONTENT INDICATORS")
    print("-" * 40)
    heading_text = []
    for heading in headings:
        text = heading.get_text(" ", strip=True)
        if text:
            heading_text.append(text)
            print("-", text)
    print("\nHYPERLINK ANALYSIS")
    print("-" * 40)
    internal_links = 0
    external_links = 0
    for link in links:
        href = link.get("href", "")
        if href.startswith("http"):
            if "example.com" in href:
                internal_links += 1
            else:
                external_links += 1
    print("Internal Links :", internal_links)
    print("External Links :", external_links)
    print("\nLINK TEXT / CONTENT ANALYSIS")
    print("-" * 40)
    link_texts = []
    for link in links:
        text = link.get_text(" ", strip=True)
        if text:
            link_texts.append(text)
    for text in link_texts[:10]:
        print("-", text)
    page_text = soup.get_text(" ", strip=True)
    words = page_text.split()
    print("\nPAGE CONTENT SUMMARY")
    print("-" * 40)
    print("Total words extracted :", len(words))
    word_counter = Counter(
        word.lower().strip(".,!?;:()[]{}\"'")
        for word in words
        if len(word.strip(".,!?;:()[]{}\"'")) > 3)
    print("\nMOST FREQUENT WORDS")
    print("-" * 40)
    for word, count in word_counter.most_common(10):
        print(f"{word:<20} : {count}")
    print("\nINTERPRETATION")
    print("-" * 40)
    print("The webpage was successfully accessed and analysed.")
    print("The extracted headings, links, paragraphs and word")
    print("frequency provide basic information about the site's")
    print("content structure and content-related activity.")
else:
    print("Unable to access the webpage.")
    print("HTTP Status Code:", response.status_code)
```
