```python
A = {"apple", "banana", "orange", "mango"}
B = {"banana", "orange", "grapes", "mango"}

intersection = A.intersection(B)

union = A.union(B)

jaccard = len(intersection) / len(union) if union else 1.0

print("Set A:", A)
print("Set B:", B)
print("Intersection:", intersection)
print("Union:", union)
print("Intersection Size:", len(intersection))
print("Union Size:", len(union))
print("Jaccard Similarity:", round(jaccard, 4))
print("Similarity Percentage:", round(jaccard * 100, 2), "%")
```
