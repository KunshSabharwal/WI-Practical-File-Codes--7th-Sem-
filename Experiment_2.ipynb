```python
import numpy as np

# Web user data
# [Pages Visited, Time Spent, Clicks]

user1 = np.array([5, 20, 10])
user2 = np.array([6, 22, 12])

# Minimum and maximum values of each feature
minimum = np.array([4, 18, 8])
maximum = np.array([15, 60, 30])

# Min-Max feature scaling
scaled1 = (user1 - minimum) / (maximum - minimum)
scaled2 = (user2 - minimum) / (maximum - minimum)

# Calculate Euclidean distance
distance = np.sqrt(np.sum((scaled1 - scaled2) ** 2))

# Calculate Euclidean similarity
similarity = 1 / (1 + distance)

print("EUCLIDEAN SIMILARITY RESULT")
print()

print("User 1:", user1)
print("User 2:", user2)

print("Scaled User 1:", np.round(scaled1, 4))
print("Scaled User 2:", np.round(scaled2, 4))

print("Euclidean Distance:", round(distance, 4))
print("Euclidean Similarity:", round(similarity, 4))
```
